# Spiking Neural Network for Speech Recognition

This notebook demonstrates how to train a Spiking Neural Network (SNN) for speech command recognition using the Speech Commands dataset.

```python
import torch
from src.model import SNN
from src.data import get_dataloaders
from src.train import train_model, test_single_sample
from src.config import NUM_EPOCHS, BATCH_SIZE, DEVICE
```

## 1. Prepare the Data

First, we'll create data loaders for training and validation sets.

```python
# Create data loaders
train_loader, val_loader = get_dataloaders(
    root='./data',  # Change this to your data directory
    batch_size=BATCH_SIZE
)
```

## 2. Create the Model

Next, we'll create an instance of our SNN model.

```python
# Create model
model = SNN(num_classes=12)  # 12 classes for speech commands
model = model.to(DEVICE)
```

## 3. Train the Model

Now we'll train the model using our training utilities.

```python
# Train the model
train_model(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    num_epochs=NUM_EPOCHS,
    device=DEVICE
)
```

## 4. Load Best Model and Evaluate

Finally, we'll load the best model and evaluate it on the validation set.

```python
# Load best model
model.load_state_dict(torch.load('best_model.pth'))

# Evaluate
from src.train import evaluate
import torch.nn as nn

criterion = nn.CrossEntropyLoss()
val_loss, val_acc = evaluate(model, val_loader, criterion, DEVICE)
print(f'Validation Loss: {val_loss:.4f}')
print(f'Validation Accuracy: {val_acc:.4f}')
```

## 5. Test on a Single Sample

Let's test the model on a single audio file from the test set.

```python
# Test on a single sample
# Replace with the path to an audio file in your test set
audio_file = './data/speech_commands_v0.02/yes/0a7c2a8d_nohash_0.wav'

# Test the model
predicted_label, confidence = test_single_sample(
    model=model,
    audio_file=audio_file,
    device=DEVICE
)

print(f"\nPrediction: {predicted_label} (confidence: {confidence:.4f})")
```
